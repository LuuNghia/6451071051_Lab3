# Bước 1: Khai báo các thư viện vào bài làm

```python
# In thông tin sinh viên
print("Họ và tên: Lưu Đình Nghĩa")
print("Mã sinh viên: 6451071051")
print("-" * 30)

from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
from PIL import Image
import numpy as np
%matplotlib inline
```

#Bước 2: Đọc vào file ảnh cần giảm kích thước và hiển thị ra màn hình

```python
# In thông tin sinh viên
print("Họ và tên: Lưu Đình Nghĩa")
print("Mã sinh viên: 6451071051")
print("-" * 30)

image = Image.open('dog.png')
axes = plt.axes(xticks=[], yticks=[])
axes.imshow(image)
plt.show()
```

# Bước 3: Chuyển ảnh đọc được thành mảng số và lấy thông tin chiều rộng, cao của ảnh

```python
# In thông tin sinh viên
print("Họ và tên: Lưu Đình Nghĩa")
print("Mã sinh viên: 6451071051")
print("-" * 30)

imarr = np.array(image)
rows = imarr.shape[0]
cols = imarr.shape[1]

print(f"Chiều cao (rows): {rows}, Chiều rộng (cols): {cols}")
```

# Bước 4: Biến đổi các giá trị màu từ [0...255] về [0...1] và chuyển dạng mảng từ 3 chiều sang 2 chiều

```python
# In thông tin sinh viên
print("Họ và tên: Lưu Đình Nghĩa")
print("Mã sinh viên: 6451071051")
print("-" * 30)

from PIL import Image
import numpy as np

# Đọc lại ảnh, ép sang RGB và lấy lại thông tin rows, cols trực tiếp tại đây
image = Image.open('dog.png').convert("RGB")
imarr = np.array(image)
rows, cols, _ = imarr.shape

# Chuẩn hóa và reshape
data = imarr / 255.0  # use 0...1 scale
data = data.reshape(rows * cols, 3)

print("Kích thước mảng sau khi reshape:", data.shape)
```

# Bước 5: Chạy thuật toán gom cụm với số màu mong muốn là 32

```python
# In thông tin sinh viên
print("Họ và tên: Lưu Đình Nghĩa")
print("Mã sinh viên: 6451071051")
print("-" * 30)

kmeans = KMeans(n_clusters=32, max_iter=300, random_state=42, n_init=10)
kmeans.fit(data)
new_colors = kmeans.cluster_centers_[kmeans.predict(data)]
```

# Bước 6: Khôi phục dạng mảng vừa thu được từ 2 chiều thành 3 chiều như ban đầu

```python
# In thông tin sinh viên
print("Họ và tên: Lưu Đình Nghĩa")
print("Mã sinh viên: 6451071051")
print("-" * 30)

img_recolored = new_colors.reshape(imarr.shape)
print("Kích thước ảnh sau khi khôi phục shape:", img_recolored.shape)
```

#Bước 7: Biểu diễn hình kết quả và so sánh với ảnh gốc ban đầu

```python
# In thông tin sinh viên
print("Họ và tên: Lưu Đình Nghĩa")
print("Mã sinh viên: 6451071051")
print("-" * 30)

fig, ax = plt.subplots(
    1, 2, figsize=(16, 6), subplot_kw=dict(xticks=[], yticks=[])
)
fig.subplots_adjust(wspace=0.05)

ax[0].imshow(image)
ax[0].set_title("Original Image", size=16)

ax[1].imshow(img_recolored)
ax[1].set_title("32-color Image", size=16)

plt.show()
```

# Bước 8: Lưu hình kết quả xuống đĩa cứng

```python
# In thông tin sinh viên
print("Họ và tên: Lưu Đình Nghĩa")
print("Mã sinh viên: 6451071051")
print("-" * 30)

img_recolored_save = img_recolored * 255
new_img = Image.fromarray(img_recolored_save.astype(np.uint8))
new_img.save("compressed_img2.png")

print("Đã lưu ảnh nén thành công với tên 'compressed_img2.png'!")
```
