### a) Đọc bộ dữ liệu Iris từ sklearn

```python
# In thông tin sinh viên
print("Họ và tên: Lưu Đình Nghĩa")
print("Mã sinh viên: 6451071051")
print("-" * 30)

from sklearn.datasets import load_iris

# Đọc bộ dữ liệu Iris từ sklearn
iris = load_iris()
data = iris.data
target = iris.target
```

### b) Chuẩn hóa dữ liệu data về đoạn [0,1]

```python
# In thông tin sinh viên
print("Họ và tên: Lưu Đình Nghĩa")
print("Mã sinh viên: 6451071051")
print("-" * 30)

import numpy as np

# Chuẩn hóa data bằng cách chia cho giá trị lớn nhất của từng cột
X_max = np.max(data, axis=0)
data_normalized = data / X_max

X = data_normalized
Y = target
```

### c) Định nghĩa hàm create_model

```python
# In thông tin sinh viên
print("Họ và tên: Lưu Đình Nghĩa")
print("Mã sinh viên: 6451071051")
print("-" * 30)

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

def create_model():
    model = Sequential([
        Dense(10, activation='relu', input_shape=(X.shape[1],)), # Hidden layer 1: 10 neuron, Relu
        Dense(20, activation='relu'),                           # Hidden layer 2: 20 neuron, Relu
        Dense(3, activation='softmax')                          # Output layer: 3 class, Softmax
    ])
    model.compile(
        optimizer='adam',
        loss='sparse_categorical_crossentropy',
        metrics=['accuracy']
    )
    return model
```

### d) Huấn luyện mô hình bằng k-fold cross validation (với k=10) và lưu kết quả từng fold

```python
# In thông tin sinh viên
print("Họ và tên: Lưu Đình Nghĩa")
print("Mã sinh viên: 6451071051")
print("-" * 30)

from sklearn.model_selection import KFold
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

# Khởi tạo KFold với n_splits=10, shuffle=True, random_state=42
kfold = KFold(n_splits=10, shuffle=True, random_state=42)

models = []
fold_metrics = []

fold_no = 1
for train_index, val_index in kfold.split(X):
    print(f'--- Đang huấn luyện Fold {fold_no}/10 ---')

    X_train_fold, X_val_fold = X[train_index], X[val_index]
    y_train_fold, y_val_fold = Y[train_index], Y[val_index]

    # Tạo và huấn luyện mô hình
    model = create_model()
    model.fit(X_train_fold, y_train_fold, epochs=100, batch_size=1, verbose=0) # verbose=0 để ẩn log chạy từng epoch cho gọn

    # Lưu mô hình
    models.append(model)

    # Dự đoán trên tập validation của fold này
    y_pred_prob = model.predict(X_val_fold, verbose=0)
    y_pred = np.argmax(y_pred_prob, axis=1)

    # Tính toán các metrics cho fold này (dùng average='macro' hoặc 'weighted')
    acc = accuracy_score(y_val_fold, y_pred)
    prec = precision_score(y_val_fold, y_pred, average='macro', zero_division=0)
    rec = recall_score(y_val_fold, y_pred, average='macro', zero_division=0)
    f1 = f1_score(y_val_fold, y_pred, average='macro', zero_division=0)

    # Tính metrics riêng cho từng class để phục vụ câu f)
    prec_per_class = precision_score(y_val_fold, y_pred, average=None, zero_division=0)
    rec_per_class = recall_score(y_val_fold, y_pred, average=None, zero_division=0)
    f1_per_class = f1_score(y_val_fold, y_pred, average=None, zero_division=0)

    fold_metrics.append({
        'fold': fold_no,
        'accuracy': acc,
        'precision': prec,
        'recall': rec,
        'f1': f1,
        'precision_class': prec_per_class,
        'recall_class': rec_per_class,
        'f1_class': f1_per_class
    })

    print(f'Fold {fold_no} -> Acc: {acc:.4f} | Prec: {prec:.4f} | Rec: {rec:.4f} | F1: {f1:.4f}')
    fold_no += 1
```

### e) In ra Average Accuracy, Average Recall, Average Precision, Average F1-score của mô hình sau k-fold cross validation

```python
# In thông tin sinh viên
print("Họ và tên: Lưu Đình Nghĩa")
print("Mã sinh viên: 6451071051")
print("-" * 30)

avg_acc = np.mean([m['accuracy'] for m in fold_metrics])
avg_prec = np.mean([m['precision'] for m in fold_metrics])
avg_rec = np.mean([m['recall'] for m in fold_metrics])
avg_f1 = np.mean([m['f1'] for m in fold_metrics])

print("=== TRUNG BÌNH CÁC CHỈ SỐ SAU K-FOLD CROSS VALIDATION ===")
print(f"Average Accuracy:  {avg_acc:.4f}")
print(f"Average Precision: {avg_prec:.4f}")
print(f"Average Recall:    {avg_rec:.4f}")
print(f"Average F1-score:  {avg_f1:.4f}")
```

### f) In ra Average Recall, Average Precision, Average F1-score cho từng class của mô hình

```python
# In thông tin sinh viên
print("Họ và tên: Lưu Đình Nghĩa")
print("Mã sinh viên: 6451071051")
print("-" * 30)

# Gom nhóm giá trị của từng class qua các fold
all_prec_class = np.array([m['precision_class'] for m in fold_metrics])
all_rec_class = np.array([m['recall_class'] for m in fold_metrics])
all_f1_class = np.array([m['f1_class'] for m in fold_metrics])

avg_prec_class = np.mean(all_prec_class, axis=0)
avg_rec_class = np.mean(all_rec_class, axis=0)
avg_f1_class = np.mean(all_f1_class, axis=0)

print("=== TRUNG BÌNH CÁC CHỈ SỐ CHO TỪNG CLASS ===")
for idx, class_name in enumerate(iris.target_names):
    print(f"Lớp [{class_name}] (Class {idx}):")
    print(f"  - Average Precision: {avg_prec_class[idx]:.4f}")
    print(f"  - Average Recall:    {avg_rec_class[idx]:.4f}")
    print(f"  - Average F1-score:  {avg_f1_class[idx]:.4f}")
```

### g) Sử dụng mô hình có F-score tốt nhất đã huấn luyện ở trên cho 03 mẫu dữ liệu mới

```python
# In thông tin sinh viên
print("Họ và tên: Lưu Đình Nghĩa")
print("Mã sinh viên: 6451071051")
print("-" * 30)

# Tìm mô hình có F1-score trung bình (macro) tốt nhất qua các fold
best_fold_idx = np.argmax([m['f1'] for m in fold_metrics])
best_model = models[best_fold_idx]
print(f"Mô hình được chọn là mô hình ở Fold {best_fold_idx + 1} với F1-score: {fold_metrics[best_fold_idx]['f1']:.4f}")

# 03 mẫu dữ liệu mới theo yêu cầu đề bài
new_samples = np.array([
    [6.2, 2.9, 4.3, 1.3],
    [5.1, 3.5, 1.4, 0.2],
    [7.3, 2.8, 6.4, 2.1]
])

# Chuẩn hóa 3 mẫu dữ liệu mới tương tự như tập train
new_samples_normalized = new_samples / X_max

# Dự đoán phân lớp
pred_prob = best_model.predict(new_samples_normalized, verbose=0)
pred_classes = np.argmax(pred_prob, axis=1)

print('\n--- KẾT QUẢ DỰ ĐOÁN CHO 3 MẪU MỚI ---')
for i, sample in enumerate(new_samples):
    cls_idx = pred_classes[i]
    cls_name = iris.target_names[cls_idx]
    print(f"Mẫu {i+1} {list(sample)} -> Nhãn lớp: {cls_idx} | Tên loài: {cls_name}")
```
