# a) Đọc bộ dữ liệu Iris từ sklearn

```python
# In thông tin sinh viên
print("Họ và tên: Lưu Đình Nghĩa")
print("Mã sinh viên: 6451071051")
print("-" * 30)

from sklearn.datasets import load_iris

# Đọc bộ dữ liệu Iris
iris = load_iris()
data = iris.data
target = iris.target
```

#b) Khảo sát bộ dữ liệu Iris

```python
# In thông tin sinh viên
print("Họ và tên: Lưu Đình Nghĩa")
print("Mã sinh viên: 6451071051")
print("-" * 30)

import pandas as pd

# 1. In ra tên các thuộc tính và thuộc tính phân lớp
print("Tên các thuộc tính (feature_names):", iris.feature_names)
print("Tên các lớp phân lớp (target_names):", iris.target_names)

# 2. In ra số lượng mẫu và số lượng thuộc tính
print("Số lượng mẫu và thuộc tính của data (shape):", data.shape)
print("Số lượng mẫu của target (shape):", target.shape)

# 3. In ra 5 mẫu dữ liệu đầu tiên
print("\n5 mẫu dữ liệu đầu tiên (data):")
print(data[:5])
print("5 nhãn đầu tiên (target):", target[:5])

# 4. Thống kê count, mean, std, min, max, tứ vị phân của bộ dữ liệu
df_iris = pd.DataFrame(data, columns=iris.feature_names)
df_iris['target'] = target
print("\nThống kê mô tả bộ dữ liệu (describe):")
display(df_iris.describe())
```

#c) Chuẩn hóa dữ liệu data về đoạn [0, 1]

```python
# In thông tin sinh viên
print("Họ và tên: Lưu Đình Nghĩa")
print("Mã sinh viên: 6451071051")
print("-" * 30)
import numpy as np
X = data
Y = target

# Chia mỗi giá trị của X cho giá trị lớn nhất trong cột tương ứng
X_max = np.max(X, axis=0)
X_normalized = X / X_max

print("Dữ liệu sau khi chuẩn hóa (5 dòng đầu):")
print(X_normalized[:5])
```

#d) Chia dữ liệu thành tập train và tập test theo tỷ lệ 8:2

```python
# In thông tin sinh viên
print("Họ và tên: Lưu Đình Nghĩa")
print("Mã sinh viên: 6451071051")
print("-" * 30)
from sklearn.model_selection import train_test_split

# Chia tập dữ liệu với test_size = 0.2 (tỷ lệ 8:2)
X_train, X_test, y_train, y_test = train_test_split(
    X_normalized, Y, test_size=0.2, random_state=42
)

print("Kích thước X_train:", X_train.shape)
print("Kích thước X_test:", X_test.shape)
```

#e) Sử dụng Cây quyết định (Decision Tree) để phân lớp

```python
# In thông tin sinh viên
print("Họ và tên: Lưu Đình Nghĩa")
print("Mã sinh viên: 6451071051")
print("-" * 30)
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from sklearn.tree import DecisionTreeClassifier
import matplotlib.pyplot as plt
import seaborn as sns


# 1. Huấn luyện mô hình Cây quyết định
dt_clf = DecisionTreeClassifier(random_state=42)
dt_clf.fit(X_train, y_train)

# 2. Dự đoán nhãn lớp cho tập test
y_pred_dt = dt_clf.predict(X_test)

# 3. Tính và in ra Accuracy
acc_dt = accuracy_score(y_test, y_pred_dt)
print(f"Decision Tree Accuracy: {acc_dt:.4f}")

# 4. Tính và in ra Precision, Recall, F1-score
print("\nDecision Tree Classification Report:")
print(classification_report(y_test, y_pred_dt, target_names=iris.target_names))

# 5. Hiển thị confusion matrix bằng heat map
plt.figure(figsize=(6, 4))
sns.heatmap(
    confusion_matrix(y_test, y_pred_dt),
    annot=True,
    fmt='d',
    cmap='Blues',
    xticklabels=iris.target_names,
    yticklabels=iris.target_names,
)
plt.title('Confusion Matrix - Decision Tree')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.show()
```

#f) Sử dụng Naïve Bayes để phân lớp

```python
# In thông tin sinh viên
print("Họ và tên: Lưu Đình Nghĩa")
print("Mã sinh viên: 6451071051")
print("-" * 30)
from sklearn.naive_bayes import GaussianNB


# 1. Huấn luyện mô hình Naïve Bayes
nb_clf = GaussianNB()
nb_clf.fit(X_train, y_train)

# 2. Dự đoán nhãn lớp cho tập test
y_pred_nb = nb_clf.predict(X_test)

# 3. Tính và in ra Accuracy
acc_nb = accuracy_score(y_test, y_pred_nb)
print(f"Naïve Bayes Accuracy: {acc_nb:.4f}")

# 4. Tính và in ra Precision, Recall, F1-score
print("\nNaïve Bayes Classification Report:")
print(classification_report(y_test, y_pred_nb, target_names=iris.target_names))

# 5. Hiển thị confusion matrix bằng heat map
plt.figure(figsize=(6, 4))
sns.heatmap(
    confusion_matrix(y_test, y_pred_nb),
    annot=True,
    fmt='d',
    cmap='Greens',
    xticklabels=iris.target_names,
    yticklabels=iris.target_names,
)
plt.title('Confusion Matrix - Naïve Bayes')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.show()
```

#g) Sử dụng K Nearest Neighbors Classifier để phân lớp

```python
# In thông tin sinh viên
print("Họ và tên: Lưu Đình Nghĩa")
print("Mã sinh viên: 6451071051")
print("-" * 30)

from sklearn.neighbors import KNeighborsClassifier

# 1. Huấn luyện mô hình KNN
knn_clf = KNeighborsClassifier(n_neighbors=3)
knn_clf.fit(X_train, y_train)

# 2. Dự đoán nhãn lớp cho tập test
y_pred_knn = knn_clf.predict(X_test)

# 3. Tính và in ra Accuracy
acc_knn = accuracy_score(y_test, y_pred_knn)
print(f"KNN Accuracy: {acc_knn:.4f}")

# 4. Tính và in ra Precision, Recall, F1-score
print("\nKNN Classification Report:")
print(classification_report(y_test, y_pred_knn, target_names=iris.target_names))

# 5. Hiển thị confusion matrix bằng heat map
plt.figure(figsize=(6, 4))
sns.heatmap(
    confusion_matrix(y_test, y_pred_knn),
    annot=True,
    fmt='d',
    cmap='Purples',
    xticklabels=iris.target_names,
    yticklabels=iris.target_names,
)
plt.title('Confusion Matrix - KNN')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.show()
```

#h) Sử dụng Neural Network để phân lớp

```python
# In thông tin sinh viên
print("Họ và tên: Lưu Đình Nghĩa")
print("Mã sinh viên: 6451071051")
print("-" * 30)

from tensorflow.keras.layers import Dense
from tensorflow.keras.models import Sequential
import tensorflow as tf


# 1. Xây dựng Neural Network gồm 02 hidden layers và 01 output layer
nn_model = Sequential([
    Dense(10, activation='relu', input_shape=(X_train.shape[1],)),  # Hidden layer 1: 10 neuron, Relu
    Dense(20, activation='relu'),  # Hidden layer 2: 20 neuron, Relu
    Dense(3, activation='softmax'),  # Output layer: 3 class (0, 1, 2), Softmax
])

# 2. Biên dịch mô hình với tối ưu hóa adam và hàm mất mát sparse_categorical_crossentropy
nn_model.compile(
    optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy']
)

# 3. Huấn luyện mô hình (epoch = 100, batch_size = 1, verbose = 1)
history = nn_model.fit(X_train, y_train, epochs=100, batch_size=1, verbose=1)

# 4. Dự đoán nhãn lớp cho tập test
y_pred_prob = nn_model.predict(X_test)
y_pred_nn = np.argmax(y_pred_prob, axis=1)

# 5. Tính và in ra Accuracy
acc_nn = accuracy_score(y_test, y_pred_nn)
print(f'\nNeural Network Accuracy: {acc_nn:.4f}')

# 6. Tính và in ra Precision, Recall, F1-score
print('\nNeural Network Classification Report:')
print(classification_report(y_test, y_pred_nn, target_names=iris.target_names))

# 7. Hiển thị confusion matrix bằng heat map
plt.figure(figsize=(6, 4))
sns.heatmap(
    confusion_matrix(y_test, y_pred_nn),
    annot=True,
    fmt='d',
    cmap='Oranges',
    xticklabels=iris.target_names,
    yticklabels=iris.target_names,
)
plt.title('Confusion Matrix - Neural Network')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.show()
```

#j) Nhận xét và đánh giá hiệu suất tổng quan
- **Đánh giá chung**: Bộ dữ liệu Iris là một bộ dữ liệu tương đối nhỏ và các thuộc tính phân tách lớp khá rõ ràng, dẫn đến hầu hết các mô hình đều đạt được hiệu suất (Accuracy) rất cao (thường từ 95% đến 100%).
- **So sánh mô hình**:
  1. **Decision Tree (Cây quyết định)**: Có tốc độ huấn luyện nhanh, dễ hiểu, tuy nhiên dễ bị overfit nếu cây quá sâu. Trên tập test nhỏ này, mô hình vẫn cho kết quả tối ưu.
  2. **Gaussian Naïve Bayes**: Hoạt động cực kỳ hiệu quả dựa trên xác suất giả định các thuộc tính độc lập. Kết quả dự đoán rất ổn định và tốn ít tài nguyên tính toán.
  3. **K-Nearest Neighbors (KNN, k=3)**: Thuật toán dựa trên khoảng cách hoạt động rất tốt đối với các không gian thuộc tính liên tục có kích thước nhỏ như Iris.
  4. **Neural Network (Mạng nơ-ron)**: Mặc dù mạnh mẽ nhất trong các kiến trúc phức tạp, mạng nơ-ron có thể hơi dư thừa (overkill) đối với bộ dữ liệu đơn giản như Iris, nhưng vẫn đạt độ chính xác gần như tuyệt đối sau 100 epochs.
- **Kết luận**: Đối với các bài toán quy mô tương tự Iris, các thuật toán học máy truyền thống như KNN hay Naïve Bayes thường được ưu tiên hơn do tính nhanh gọn và không cần cấu hình siêu tham số phức tạp như mạng nơ-ron.
